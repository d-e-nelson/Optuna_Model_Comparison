```
import pandas as pd
import numpy as np
```

```
df = pd.read_csv("kc_house_data.csv")
```

## Example Dataset

```
df.head()
```

```
print(df.isna().sum().sum())
print(df.isna().sum())
```

```
negitive_df = df.select_dtypes(include=[np.number])

negitive_count = (
    (negitive_df <0)
    .sum()
    .loc[lambda x: x > 0]
    .sort_values(ascending = False)
)

print(negitive_count)
```

```
place_holder_columns = (
    df.isin([-9999,-8888, -7777,-6666,-5555,-4444,-3333,-2222,-1111,-100])
    .sum()
    .loc[lambda x : x>0]
    .sort_values(ascending = False)
)

print(place_holder_columns)
```

```
df['date'] = pd.to_datetime(df['date'])
```

```
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df = df.drop(columns=['date'])
```

### Examine Bedrooms

```
df['bedrooms'].value_counts().sort_index()
```

```
Q1 = df['bedrooms'].quantile(0.25)
Q3 = df["bedrooms"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(lower_bound, upper_bound)
```

```
df = df[(df['bedrooms'] > 0) & (df['bedrooms'] <= 10)]
```

### Examine Bathrooms

```
df['bathrooms'].value_counts().sort_index()
```

```
Q1 = df['bathrooms'].quantile(0.25)
Q3 = df["bathrooms"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(lower_bound, upper_bound)
```

```
df = df[(df['bathrooms'] >= 1) & (df['bathrooms'] <= 6)]
```

### Examine sqft\_living

```
pd.set_option('display.max_rows',2000)
df['sqft_living'].value_counts().sort_index()
```

```
df['sqft_living'].describe()
```

```
Q1 = df['sqft_living'].quantile(0.25)
Q3 = df["sqft_living"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(lower_bound, upper_bound)
```

```
df['sqft_living'] = np.log1p(df['sqft_living'])
```

```
df['sqft_living'].describe()
```

### Examine: Price

```
df['price'].value_counts().sort_index()
```

```
df['price'].describe()
```

```
df['price'] = np.log1p(df['price'])
```

```
df['price'].describe()
```

### Examine: sqft\_lot

```
df['sqft_lot'].value_counts().sort_index()
```

```
df['sqft_lot'].describe()
```

```
df['sqft_lot'] = df['sqft_lot'].clip(upper=100000)
```

```
df['sqft_lot'] = np.log1p(df['sqft_lot'])
```

### Examine: sqft\_living15

```
df['sqft_living15'].value_counts().sort_index()
```

```
df['sqft_living15'].describe()
```

```
df['sqft_living15'] = np.log1p(df['sqft_living15'])
```

### Examine: sqft\_lot15

```
df['sqft_lot15'].value_counts().sort_index()
```

```
df['sqft_lot15'].describe()
```

```
df['sqft_lot15'] = np.log1p(df['sqft_lot15'])
```

### Decided Which Columns to Drop

```
df.columns
```

```
df = df.drop(columns=['id'])
```

```
df = df.drop(columns=['sqft_above'])
```

```
df['zipcode'].nunique()
```

```
df['zipcode'].value_counts().head()
```

```
df = pd.get_dummies(df, columns=['zipcode'], drop_first=True)
```

### Add Additional Columns

```
df['house_age'] = df['year'] - df['yr_built']
```

```
df['was_renovated'] = (df['yr_renovated'] > 0).astype(int)
```

```
df['sqft_ratio'] = df['sqft_living'] / df['sqft_lot']
```

```
df['bath_per_bed'] = df['bathrooms'] / df['bedrooms']
```
